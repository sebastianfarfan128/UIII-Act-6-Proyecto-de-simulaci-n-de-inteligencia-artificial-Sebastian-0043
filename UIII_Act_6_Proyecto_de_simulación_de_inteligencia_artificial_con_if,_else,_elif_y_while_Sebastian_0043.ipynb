### Proyecto de simulación de inteligencia artificial con if, else, elif y while
### 18. Interés compuesto

```python
print("--- Asistente de IA: Calculadora de Interés Compuesto ---")
capital = float(input("¿Cuál es tu capital inicial? "))
tasa = float(input("¿Cuál es la tasa de interés anual (ej: 0.05 para 5%)? "))
años = int(input("¿A cuántos años es la inversión? "))

contador = 1
while contador <= años:
    interes = capital * tasa
    capital = capital + interes
    print("Año " + str(contador) + ": $" + str(round(capital, 2)))
    contador = contador + 1

if capital > 1000000:
    print("¡IA detecta un gran crecimiento! Serás millonario.")
elif capital > 0:
    print("IA completa: Tu dinero está creciendo de forma constante.")
else:
    print("Alerta: Revisa los valores ingresados.")
print("Programa elaborado por Sebastian Farfan 0043 NL = 18")
```
